```python

#Lab 3-4: Harris Corner Detection
#Image Processing

#Import needed libraries
import os
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

#Parameters
IMAGE_PATH = "/content/harris_corner_detection_image.jpg"
BLOCK_SIZE = 2
APERTURE_SIZE = 3
K = 0.02
THRESH_RATIO = 0.04

#Load image
image = cv2.imread(IMAGE_PATH)

original = image.copy()
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

harris = cv2.cornerHarris(gray, BLOCK_SIZE, APERTURE_SIZE, K)

harris = cv2.dilate(harris, None)

threshold = THRESH_RATIO * harris.max()
ys, xs = np.where(harris > threshold)

# Draw circles around detected corners on the original image
for x, y in zip(xs, ys):
    cv2.circle(image, (int(x), int(y)), 1, (0, 0, 255), -1)

print(f"Detected corner pixels: {len(xs)}")

#Display
print("Original Image:")
cv2_imshow(original)

print("Harris Corners Detected:")
cv2_imshow(image)
```
